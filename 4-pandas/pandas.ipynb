Como esse arquivo que escolhi em especifico está no formato de relatório do excel e não uma tabela analítica, devemos fazer algumas alterações

```python
raw = pd.read_excel("/home/felipe/Desktop/data_engineering_journey/data/producao-petroleo-b.xls", header=None)

print(raw.shape)
raw.head(40)
```

Podemos fazer dessa forma para identificar apenas onde estão os dados

```python
for i, row in raw.iterrows():
    if row.notna().sum() > 5:
        print(i)
```

Perceba o salto de 44 para 93, da pra entender que temos nessa mesma planilha duas tabelas  

Extraindo apenas a primeira tabela

```python
petroleo = raw.iloc[31:45].copy()

petroleo.head()
```

transformando agora a primeira linha em cabeçalho

```python
petroleo.columns = petroleo.iloc[0]

petroleo = petroleo.iloc[1:]

petroleo.head()
```

```python
## verificando os tipos

petroleo.info()
```

### Exploração inicial agora

vendo se há agora nessa "nova" tabela petroleo, valores faltantes

```python
petroleo.isna().sum()
```

Fazendo um resumo estatístico

```python
petroleo.describe()
```

Acontece que agora os dados estão nesse formato de Mês | 2000 | 2001 | 2002 | ... mas da pra fazer uma análise melhor se converter para o formato de tidy data. vamos em outras palavras, alinhar Mẽs | ano | produção

```python
print(petroleo.columns.tolist())
```

```python
petroleo_long = pd.melt(
    petroleo,
    id_vars="Mês",
    var_name="Ano",
    value_name="Produção"
)

petroleo_long.head()
```
